```python
history = []
count = 0
while True :
    print("---------SIMPLE CALCULATOR---------")
    print("1.  ADD")
    print("2.  SUBTRACT")
    print("3.  MULTIPLY")
    print("4.  DIVIDE")
    print("5.  EXIT")
    print("6.  POWER")
    print("7.  MODULUS")
    print("8.  SHOW HISTORY")
    print("9.  CLEAR HISTORY")
    print("10. CLEAR HISTORY and RESET COUNTER")
    
    choice = int(input("Enter your choice(1 - 10) : "))


    if choice == 5 :
        print("Calculator closed")
        break

    elif choice == 8 :
        if len(history) == 0 :
            print("No history available")
            continue

        print("-----HISTORY-----")
        for item in history:
            print(item)
        continue
        
    elif choice == 9 :
        history.clear()
        print("History cleared successfully")
        continue

    elif choice == 10 :
        history.clear()
        count = 0
        print("History cleared successfully and counter is reset")
        continue
        
    elif choice not in [1,2,3,4,6,7,] :
        print("Invalid Choice")
        continue
     

    num1 = int(input("Enter the 1st  number :"))
    num2 = int(input("Enter the 2nd  number :"))
    
    if choice == 1 :
        res = num1 + num2
        history.append(f"{num1} + {num2} = {res}")
        print(f"Result = {res}")
        count  +=  1
        
    elif choice == 2 :
        res = num1 - num2
        history.append(f"{num1} - {num2} = {res}")
        print(f"Result = {res}")
        count  +=  1
    
    elif choice == 3 :
        res = num1 * num2
        history.append(f"{num1} * {num2} = {res}")
        print(f"Result = {res}")
        count  +=  1

    elif choice == 4 :
        if num2 == 0:
            print("Not possible")
            continue
        res = num1 // num2
        history.append(f"{num1} // {num2} = {res}")
        print(f"Result = {res}")
        count  +=  1
    
    elif choice == 6 :
        res = num1 ** num2
        history.append(f"{num1} ** {num2} = {res}")
        print(f"Result = {res}")
        count  +=  1
        
    elif choice == 7 :
        if num2 == 0:
            print("Not possible")
            continue
        res = num1 % num2
        history.append(f"{num1} % {num2} = {res}")
        print(f"Result = {res}")
        count  += 1

print(f"-----------Total Calculation Performed : {count}-----------")
    
```
